## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Creating a copy to preserve the original dataset
df_clean = df.copy()

# removing unwanted spaces
df_clean.columns = df_clean.columns.str.strip()

# Removing leading and trailing spaces from text columns
text_columns = df_clean.select_dtypes(include="object").columns

for column in text_columns:
    df_clean[column] = df_clean[column].astype(str).str.strip()

# Removing duplicate records if any
duplicates_before = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates().copy()

# Converting Credit_Score into an ordered categorical variable for meaningful analysis
credit_score_order = ["Poor", "Standard", "Good"]

df_clean["Credit_Score"] = pd.Categorical(
    df_clean["Credit_Score"],
    categories=credit_score_order,
    ordered=True
)

# Checking outliers in key financial and repayment variables using the IQR method
outlier_columns = [
    "Annual_Income",
    "Monthly_Inhand_Salary",
    "Outstanding_Debt",
    "Delay_from_due_date",
    "Num_of_Delayed_Payment",
    "Num_Credit_Inquiries",
    "Credit_Utilization_Ratio"
]
outlier_summary = []

for column in outlier_columns:
    q1 = df_clean[column].quantile(0.25)
    q3 = df_clean[column].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outlier_count = (
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ).sum()

    outlier_summary.append([column, lower_limit, upper_limit, outlier_count])

outlier_summary = pd.DataFrame(
    outlier_summary,
    columns=["Column", "Lower Limit", "Upper Limit", "Outlier Count"]
)

print(f"Duplicate rows removed: {duplicates_before}")
print(f"Final dataset shape: {df_clean.shape}")

display(outlier_summary)





````

### What all manipulations have you done and insights you found?

The original dataset was preserved and a working copy named `df_clean` was created for analysis. Column names and text values were standardised by removing unnecessary spaces. Duplicate rows were checked and removed only if present. The Credit_Score variable was converted into an ordered categorical variable with the order Poor, Standard, and Good so that comparisons across risk groups are meaningful.

Missing values were checked earlier and none were found. Key financial and repayment variables such as annual income, outstanding debt, delayed-payment days, delayed-payment count, credit inquiries, and credit-utilisation ratio were checked for outliers using the IQR method.

The detected outliers were not automatically removed because extreme income, debt, or repayment values can represent genuine high-risk customer behaviour in a financial dataset. Removing them without validation could hide important risk patterns. Therefore, they were retained and interpreted carefully during the visual analysis.
